##### **SpaCy method**

```python
import pandas as pd

df = pd.read_csv("data/metadata.csv")
df
```

The spaCy `en_core_web_md` model is loaded, providing pretrained word embeddings.
This allows the diagnosis text to be transformed into numerical vector representations.

```python
import spacy
import numpy as np

nlp = spacy.load("en_core_web_md")
nlp
```

Each `remark` is transformed into an embedding vector, providing a numerical representation of its semantic content
for each patient.

```python
def get_sentence_vector(text):
    doc = nlp(text)
    return doc.vector

df["text_vector"] = df["remarks"].apply(get_sentence_vector)
df
```

```python
df.to_csv("data/metadata_with_text_vectors.csv", index=False)
```
