# JOINING DATASETS

```python
df_patients.columns
```

```python
df_patients["#Patient Identifier"].value_counts()
```

```python
df_sample["Patient Identifier"].value_counts()
```

```python
df_merged = df_patients.merge(
    df_sample,
    left_on="#Patient Identifier",
    right_on="Patient Identifier",
    how="inner"
)
df_merged
```

```python
# Build patient ID from mutation barcode and join with patient table
df_mutations = df_mutations.copy()
df_merged = df_merged.copy()

# Example: P-0081657-T01-IM7 -> P-0081657
df_mutations["Patient_ID"] = (
    df_mutations["Tumor_Sample_Barcode"]
    .astype(str)
    .str.extract(r"^(P-\d+)", expand=False)
    .str.strip()
 )

df_merged["Patient_ID"] = df_merged["#Patient Identifier"].astype(str).str.strip()

# Inner join keeps only matched patients; use how='left' to keep all mutations
df_joined = df_mutations.merge(
    df_merged,
    on="Patient_ID",
    how="inner",
    suffixes=("_mut", "_pat")
)

print(f"Mutations rows: {len(df_mutations):,}")
print(f"Patients rows: {len(df_merged):,}")
print(f"Joined rows: {len(df_joined):,}")

# Show a few matched keys
print(df_joined[["Tumor_Sample_Barcode", "Patient_ID", "#Patient Identifier"]].head(10))
```

```python
# filter for specific genes
Kras_filter = False
if Kras_filter:
    print(df_joined['Hugo_Symbol'].unique())
    genes_of_interest = ['KRAS']
    df_filtered = df_joined[df_joined['Hugo_Symbol'].isin(genes_of_interest)]
    print(df_filtered[['Hugo_Symbol', 'Patient_ID']])
else:
    df_filtered = df_joined.copy()
    print(df_filtered[['Hugo_Symbol', 'Patient_ID']])
```

```python
# save the dataset
df_filtered.to_csv("data_pancreas/filtered_data_PANCREAS.csv", index=False)
```
