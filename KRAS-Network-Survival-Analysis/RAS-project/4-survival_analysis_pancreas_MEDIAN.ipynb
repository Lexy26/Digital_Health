```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from lifelines import KaplanMeierFitter, CoxPHFitter
from lifelines.statistics import logrank_test


# ============================================================
# Utility function
# ============================================================

def format_p_value(p):
    """
    Format p-values for plot titles.
    """
    if pd.isna(p):
        return "NA"
    elif p < 0.001:
        return f"{p:.2e}"
    else:
        return f"{p:.3f}"


# ============================================================
# 1. Load filtered clinical + mutation data
# ============================================================

df = pd.read_csv("data_pancreas/filtered_data_PANCREAS.csv", low_memory=False)

df = df.rename(columns={
    "Overall Survival (Months)": "OS_MONTHS",
    "Overall Survival Status": "OS_STATUS",
    "Patient_ID": "PATIENT_ID",
    "Hugo_Symbol": "Gene"
})

df = df.loc[:, ~df.columns.duplicated()].copy()

df["OS_MONTHS"] = pd.to_numeric(df["OS_MONTHS"], errors="coerce")

df["event"] = (
    df["OS_STATUS"]
    .astype(str)
    .str.contains("DECEASED", case=False, na=False)
    .astype(int)
)

df = df.dropna(subset=["PATIENT_ID", "Gene", "OS_MONTHS", "event"])


# ============================================================
# 2. Load merged dataset to define valid network patients
# ============================================================

df_merged = pd.read_csv("data_pancreas/merged_mutations_cna_sv.csv", low_memory=False)
df_merged = df_merged.loc[:, ~df_merged.columns.duplicated()].copy()

if "PATIENT_ID" not in df_merged.columns:
    if "Patient_ID" in df_merged.columns:
        df_merged = df_merged.rename(columns={"Patient_ID": "PATIENT_ID"})
    elif "Tumor_Sample_Barcode" in df_merged.columns:
        df_merged["PATIENT_ID"] = (
            df_merged["Tumor_Sample_Barcode"]
            .astype(str)
            .str[:12]
        )
    elif "SAMPLE_ID" in df_merged.columns:
        df_merged["PATIENT_ID"] = (
            df_merged["SAMPLE_ID"]
            .astype(str)
            .str[:12]
        )
    else:
        raise ValueError("Non trovo una colonna paziente in merged.")

valid_patients = set(
    df_merged["PATIENT_ID"]
    .dropna()
    .astype(str)
    .unique()
)

print("Patients valid from merged/network:", len(valid_patients))


# ============================================================
# 3. Create clinical table restricted to network patients
# ============================================================

df["PATIENT_ID"] = df["PATIENT_ID"].astype(str)

clinical = (
    df[["PATIENT_ID", "OS_MONTHS", "event"]]
    .drop_duplicates("PATIENT_ID")
    .copy()
)

clinical = clinical[
    clinical["PATIENT_ID"].isin(valid_patients)
].copy()

print(
    "Patients with survival data after merged restriction:",
    clinical["PATIENT_ID"].nunique()
)


# ============================================================
# 4. Load Leiden clusters with hub ranking from CSV
# ============================================================

clusters_df = pd.read_csv("data_pancreas/leiden_clusters.csv")

clusters_df = clusters_df.rename(columns={
    "Hugo_Symbol": "Gene",
    "gene": "Gene",
    "cluster": "Cluster",
    "hubscore": "HubScore",
    "rank_in_cluster": "Rank_in_cluster"
})

clusters_df["Gene"] = clusters_df["Gene"].astype(str)
clusters_df["Cluster"] = clusters_df["Cluster"].astype(int)

# Se Rank_in_cluster non esiste, lo ricostruisco da HubScore
if "Rank_in_cluster" not in clusters_df.columns:

    if "HubScore" not in clusters_df.columns:
        raise ValueError(
            "Nel file leiden_clusters.csv non trovo né Rank_in_cluster né HubScore. "
            "Devi salvare prima il file Leiden con le metriche hub."
        )

    clusters_df["HubScore"] = pd.to_numeric(
        clusters_df["HubScore"],
        errors="coerce"
    )

    clusters_df = clusters_df.sort_values(
        ["Cluster", "HubScore"],
        ascending=[True, False]
    )

    clusters_df["Rank_in_cluster"] = (
            clusters_df
            .groupby("Cluster")
            .cumcount() + 1
    )

clusters_df["Rank_in_cluster"] = pd.to_numeric(
    clusters_df["Rank_in_cluster"],
    errors="coerce"
)

if "HubScore" in clusters_df.columns:
    clusters_df["HubScore"] = pd.to_numeric(
        clusters_df["HubScore"],
        errors="coerce"
    )

clusters_df = clusters_df.sort_values(
    ["Cluster", "Rank_in_cluster"]
).copy()


def get_top_hub_genes_text(cluster_id, max_genes=10):
    sub = clusters_df[
        clusters_df["Cluster"] == cluster_id
        ].copy()

    sub = sub.sort_values("Rank_in_cluster")

    labels = []

    for _, row in sub.head(max_genes).iterrows():
        gene = row["Gene"]
        rank = int(row["Rank_in_cluster"])

        if "HubScore" in sub.columns and pd.notna(row.get("HubScore", np.nan)):
            labels.append(f"{rank}. {gene} ({row['HubScore']:.3f})")
        else:
            labels.append(f"{rank}. {gene}")

    if len(sub) > max_genes:
        labels.append(f"... +{len(sub) - max_genes} genes")

    return " | ".join(labels)


cluster_gene_dict = (
    clusters_df
    .groupby("Cluster")["Gene"]
    .apply(list)
    .to_dict()
)

for cluster, genes in cluster_gene_dict.items():
    print(f"Cluster {cluster}: {len(genes)} genes")
    print("Top hubs:", get_top_hub_genes_text(cluster, max_genes=10))


# ============================================================
# 5. Restrict mutation data to valid network patients
# ============================================================

df_network = df[
    df["PATIENT_ID"].isin(valid_patients)
].copy()

print("Mutation rows after merged restriction:", df_network.shape[0])
print(
    "Patients in mutation table after restriction:",
    df_network["PATIENT_ID"].nunique()
)


# ============================================================
# 6. Survival analysis by Leiden cluster
# ============================================================

results = []

for cluster_id, genes in cluster_gene_dict.items():

    genes = set(genes)
    top_hub_genes_text = get_top_hub_genes_text(
        cluster_id,
        max_genes=10
    )

    patients_with_cluster = (
        df_network[
            df_network["Gene"].isin(genes)
        ]["PATIENT_ID"]
        .dropna()
        .unique()
    )

    surv_df = clinical.copy()

    surv_df["has_cluster"] = (
        surv_df["PATIENT_ID"]
        .isin(patients_with_cluster)
        .astype(int)
    )

    n_cluster = int(surv_df["has_cluster"].sum())
    n_no_cluster = int((surv_df["has_cluster"] == 0).sum())

    events_cluster = int(
        surv_df.loc[
            surv_df["has_cluster"] == 1,
            "event"
        ].sum()
    )

    events_no_cluster = int(
        surv_df.loc[
            surv_df["has_cluster"] == 0,
            "event"
        ].sum()
    )

    print("\n" + "=" * 70)
    print(f"Cluster {cluster_id}")
    print(f"Genes in cluster: {len(genes)}")
    print(f"Top hub genes: {top_hub_genes_text}")
    print(f"Patients with cluster mutation: {n_cluster}")
    print(f"Patients without cluster mutation: {n_no_cluster}")
    print(f"Events with cluster mutation: {events_cluster}")
    print(f"Events without cluster mutation: {events_no_cluster}")

    if (
            n_cluster < 10
            or n_no_cluster < 10
            or events_cluster < 5
            or events_no_cluster < 5
    ):
        print("Skipped: too few patients or events.")
        continue

    group1 = surv_df[
        surv_df["has_cluster"] == 1
        ].copy()

    group0 = surv_df[
        surv_df["has_cluster"] == 0
        ].copy()

    # ------------------------------------------------------------
    # Log-rank test
    # ------------------------------------------------------------

    lr = logrank_test(
        group1["OS_MONTHS"],
        group0["OS_MONTHS"],
        event_observed_A=group1["event"],
        event_observed_B=group0["event"]
    )

    # ------------------------------------------------------------
    # Cox proportional hazards model
    # ------------------------------------------------------------

    cox_df = surv_df[
        ["OS_MONTHS", "event", "has_cluster"]
    ].copy()

    cph = CoxPHFitter()

    cph.fit(
        cox_df,
        duration_col="OS_MONTHS",
        event_col="event"
    )

    hr = np.exp(cph.params_["has_cluster"])
    p_cox = cph.summary.loc["has_cluster", "p"]

    results.append({
        "Cluster": cluster_id,
        "Num_genes": len(genes),
        "Patients_with_cluster": n_cluster,
        "Events_with_cluster": events_cluster,
        "Patients_without_cluster": n_no_cluster,
        "Events_without_cluster": events_no_cluster,
        "Hazard_Ratio": hr,
        "Cox_p_value": p_cox,
        "Logrank_p_value": lr.p_value,
        "Top_hub_genes": top_hub_genes_text,
        "Genes": ", ".join(sorted(genes))
    })

    # ========================================================
    # Kaplan-Meier plot with statistical information above
    # ========================================================

    kmf = KaplanMeierFitter()

    fig, ax = plt.subplots(figsize=(14, 6))

    kmf.fit(
        group1["OS_MONTHS"],
        event_observed=group1["event"],
        label=f"Cluster {cluster_id} mutated"
    )
    kmf.plot_survival_function(ax=ax)

    kmf.fit(
        group0["OS_MONTHS"],
        event_observed=group0["event"],
        label=f"Cluster {cluster_id} not mutated"
    )
    kmf.plot_survival_function(ax=ax)

    main_title = (
        f"Kaplan-Meier Survival Curve - Leiden Cluster {cluster_id}"
    )

    stats_line = (
        f"HR={hr:.2f}, "
        f"Cox p={format_p_value(p_cox)}, "
        f"Log-rank p={format_p_value(lr.p_value)}"
    )

    top_hubs_line = (
        f"Top hub genes: {top_hub_genes_text}"
    )

    fig.suptitle(
        main_title,
        fontsize=11,
        y=0.98
    )

    fig.text(
        0.5,
        0.945,
        stats_line,
        ha="center",
        va="center",
        fontsize=10
    )

    fig.text(
        0.5,
        0.915,
        top_hubs_line,
        ha="center",
        va="center",
        fontsize=9
    )

    ax.set_xlabel("Overall Survival Months")
    ax.set_ylabel("Survival probability")
    ax.grid(alpha=0.3)
    ax.legend(loc="upper right")

    # spazio sopra il grafico per titolo + statistiche + top hub genes
    plt.subplots_adjust(top=0.84)

    plt.show()


# ============================================================
# 7. Save final results
# ============================================================

survival_results = pd.DataFrame(results)

if len(survival_results) > 0:

    survival_results = (
        survival_results
        .sort_values("Logrank_p_value")
        .reset_index(drop=True)
    )

    display(survival_results)

    survival_results.to_csv(
        "data_pancreas/survival_analysis_by_leiden_cluster_network_restricted.csv",
        index=False
    )

    print("Saved: survival_analysis_by_leiden_cluster_network_restricted.csv")

else:
    print("No cluster had enough patients/events for survival analysis.")
```
