## NETWORK TO CLINICAL

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from pathlib import Path
from lifelines import KaplanMeierFitter, CoxPHFitter
from lifelines.statistics import logrank_test


# ============================================================
# Helper: collapse duplicate columns safely
# ============================================================

def collapse_duplicate_columns(df):
    """
    Se dopo un rename si creano colonne duplicate, le collassa.
    Per ogni nome duplicato prende il primo valore non nullo da sinistra a destra.
    """
    df = df.copy()

    if not df.columns.duplicated().any():
        return df

    ordered_unique_cols = pd.unique(df.columns)
    collapsed = pd.DataFrame(index=df.index)

    for col in ordered_unique_cols:
        same_name_cols = df.loc[:, df.columns == col]

        if same_name_cols.shape[1] == 1:
            collapsed[col] = same_name_cols.iloc[:, 0]
        else:
            collapsed[col] = same_name_cols.bfill(axis=1).iloc[:, 0]

    return collapsed.copy()


def get_single_column(df, col):
    """
    Restituisce sempre una Series anche se per errore esistono colonne duplicate.
    """
    obj = df[col]

    if isinstance(obj, pd.DataFrame):
        return obj.iloc[:, 0]

    return obj


# ============================================================
# 0. Paths
# ============================================================

data_dir = Path("data_lung")

filtered_path = data_dir / "filtered_data_LUNG.csv"
merged_path = data_dir / "merged_mutations_cna_sv.csv"

clusters_path = (
        data_dir
        / "networks_by_oncotree_groups"
        / "final_leiden_clusters_with_hub_metrics"
        / "leiden_clusters_with_hub_metrics_all_oncotree_groups.csv"
)

survival_output_dir = (
        data_dir
        / "networks_by_oncotree_groups"
        / "survival_by_oncotree_groups"
)

survival_output_dir.mkdir(parents=True, exist_ok=True)

if not clusters_path.exists():
    raise FileNotFoundError(
        f"Non trovo il file dei cluster Leiden per gruppi Oncotree:\n{clusters_path}\n\n"
        "Controlla di aver eseguito prima la cella che crea "
        "`leiden_clusters_all_oncotree_groups.csv`."
    )


# ============================================================
# 0B. Gruppi Oncotree da analizzare
# ============================================================

oncotree_groups = {
    "LUAD_NSCLC": ["LUAD", "NSCLC"],
    "LUSC_NSCLC": ["LUSC", "NSCLC"],
}

group_labels = {
    "LUAD_NSCLC": "LUAD+NSCLC",
    "LUSC_NSCLC": "LUSC+NSCLC",
}


# ============================================================
# 1. Load filtered clinical + mutation data
# ============================================================

df = pd.read_csv(filtered_path, low_memory=False)
df = collapse_duplicate_columns(df)

# Se esiste già una colonna Gene e anche Hugo_Symbol, preferisco Hugo_Symbol
if "Gene" in df.columns and "Hugo_Symbol" in df.columns:
    df = df.drop(columns=["Gene"])

df = df.rename(columns={
    "Overall Survival (Months)": "OS_MONTHS",
    "Overall Survival Status": "OS_STATUS",
    "Patient_ID": "PATIENT_ID",
    "Hugo_Symbol": "Gene",
    "Oncotree Code": "ONCOTREE_CODE"
})

df = collapse_duplicate_columns(df)

required_filtered_cols = {
    "PATIENT_ID",
    "Gene",
    "ONCOTREE_CODE",
    "OS_MONTHS",
    "OS_STATUS"
}

missing_filtered = required_filtered_cols - set(df.columns)

if missing_filtered:
    raise ValueError(
        f"Colonne mancanti in filtered_data_LUNG.csv: {missing_filtered}"
    )

df["PATIENT_ID"] = get_single_column(df, "PATIENT_ID").astype(str)
df["Gene"] = get_single_column(df, "Gene").astype(str)
df["ONCOTREE_CODE"] = get_single_column(df, "ONCOTREE_CODE").astype(str)

df["OS_MONTHS"] = pd.to_numeric(
    get_single_column(df, "OS_MONTHS"),
    errors="coerce"
)

df["event"] = (
    get_single_column(df, "OS_STATUS")
    .astype(str)
    .str.contains("DECEASED", case=False, na=False)
    .astype(int)
)

df = df.dropna(
    subset=[
        "PATIENT_ID",
        "Gene",
        "ONCOTREE_CODE",
        "OS_MONTHS",
        "event"
    ]
).copy()

df = collapse_duplicate_columns(df)

print("Filtered clinical/mutation data loaded")
print("Rows:", df.shape[0])
print("Patients:", df["PATIENT_ID"].nunique())
print("Oncotree Codes:", sorted(df["ONCOTREE_CODE"].dropna().unique()))

print("\nDuplicate columns in df:")
print(df.columns[df.columns.duplicated()].tolist())


# ============================================================
# 2. Load merged dataset to define valid network patients
# ============================================================

df_merged = pd.read_csv(merged_path, low_memory=False)
df_merged = collapse_duplicate_columns(df_merged)

if "Gene" in df_merged.columns and "Hugo_Symbol" in df_merged.columns:
    df_merged = df_merged.drop(columns=["Gene"])

df_merged = df_merged.rename(columns={
    "Patient_ID": "PATIENT_ID",
    "Hugo_Symbol": "Gene"
})

df_merged = collapse_duplicate_columns(df_merged)

if "PATIENT_ID" not in df_merged.columns:
    if "Tumor_Sample_Barcode" in df_merged.columns:
        df_merged["PATIENT_ID"] = (
            df_merged["Tumor_Sample_Barcode"]
            .astype(str)
            .str[:12]
        )
    elif "SAMPLE_ID" in df_merged.columns:
        df_merged["PATIENT_ID"] = (
            df_merged["SAMPLE_ID"]
            .astype(str)
            .str[:12]
        )
    else:
        raise ValueError("Non trovo una colonna paziente in merged.")

if "Gene" not in df_merged.columns:
    raise ValueError(
        "Non trovo la colonna Gene/Hugo_Symbol in merged_mutations_cna_sv.csv."
    )

df_merged["PATIENT_ID"] = get_single_column(df_merged, "PATIENT_ID").astype(str)
df_merged["Gene"] = get_single_column(df_merged, "Gene").astype(str)

valid_patients_global = set(
    df_merged["PATIENT_ID"]
    .dropna()
    .astype(str)
    .unique()
)

print("\nPatients valid from merged/network:", len(valid_patients_global))

print("\nDuplicate columns in df_merged:")
print(df_merged.columns[df_merged.columns.duplicated()].tolist())


# ============================================================
# 3. Annotate merged with Oncotree Code
# ============================================================

patient_oncotree = (
    df[["PATIENT_ID", "ONCOTREE_CODE"]]
    .dropna()
    .drop_duplicates()
)

multi_oncotree = (
    patient_oncotree
    .groupby("PATIENT_ID")["ONCOTREE_CODE"]
    .nunique()
)

patients_with_multiple_codes = multi_oncotree[multi_oncotree > 1]

if len(patients_with_multiple_codes) > 0:
    print(
        "\nATTENZIONE: alcuni pazienti hanno più Oncotree Code. "
        "Verranno mantenute le coppie paziente-istologia."
    )
    print(patients_with_multiple_codes.head())
else:
    print("\nOK: ogni paziente ha un solo Oncotree Code.")

df_merged_oncotree = df_merged.merge(
    patient_oncotree,
    on="PATIENT_ID",
    how="inner"
)

df_merged_oncotree = collapse_duplicate_columns(df_merged_oncotree)

df_merged_oncotree = df_merged_oncotree.dropna(
    subset=["PATIENT_ID", "Gene", "ONCOTREE_CODE"]
).copy()

print("Merged annotated with Oncotree Code")
print("Rows:", df_merged_oncotree.shape[0])
print("Patients:", df_merged_oncotree["PATIENT_ID"].nunique())


# ============================================================
# 4. Clinical table per gruppi Oncotree
# ============================================================

clinical_base = (
    df[["PATIENT_ID", "ONCOTREE_CODE", "OS_MONTHS", "event"]]
    .drop_duplicates(["PATIENT_ID", "ONCOTREE_CODE"])
    .copy()
)

clinical_base = clinical_base[
    clinical_base["PATIENT_ID"].isin(valid_patients_global)
].copy()

clinical_group_tables = []

for group_name, codes in oncotree_groups.items():

    readable_group_name = group_labels.get(group_name, group_name)

    clinical_group = clinical_base[
        clinical_base["ONCOTREE_CODE"].isin(codes)
    ].copy()

    clinical_group["Oncotree_Group"] = group_name
    clinical_group["Included_Oncotree_Codes"] = readable_group_name

    clinical_group_tables.append(clinical_group)

clinical = pd.concat(
    clinical_group_tables,
    ignore_index=True
)

print("\nPatients with survival data after merged restriction, by Oncotree group:")

display(
    clinical
    .groupby("Oncotree_Group")
    .agg(
        included_codes=("Included_Oncotree_Codes", "first"),
        n_patients=("PATIENT_ID", "nunique"),
        n_events=("event", "sum")
    )
    .sort_values("n_patients", ascending=False)
)


# ============================================================
# 5. Load Leiden clusters group-oriented
# ============================================================

clusters_df = pd.read_csv(clusters_path)
clusters_df = collapse_duplicate_columns(clusters_df)

if "Gene" in clusters_df.columns and "Hugo_Symbol" in clusters_df.columns:
    clusters_df = clusters_df.drop(columns=["Hugo_Symbol"])

clusters_df = clusters_df.rename(columns={
    "Hugo_Symbol": "Gene",
    "gene": "Gene",
    "cluster": "Cluster",
    "Leiden_cluster": "Cluster",
    "hubscore": "HubScore",
    "rank_in_cluster": "Rank_in_cluster",
    "Oncotree_Code": "Oncotree_Group",
    "ONCOTREE_CODE": "Oncotree_Group",
    "Oncotree Code": "Oncotree_Group"
})

clusters_df = collapse_duplicate_columns(clusters_df)

required_cluster_cols = {
    "Oncotree_Group",
    "Gene",
    "Cluster"
}

missing_cluster_cols = required_cluster_cols - set(clusters_df.columns)

if missing_cluster_cols:
    raise ValueError(
        f"Colonne mancanti nel file cluster Leiden per gruppi: {missing_cluster_cols}"
    )

clusters_df["Oncotree_Group"] = get_single_column(
    clusters_df,
    "Oncotree_Group"
).astype(str)

clusters_df["Gene"] = get_single_column(
    clusters_df,
    "Gene"
).astype(str)

clusters_df["Cluster"] = pd.to_numeric(
    get_single_column(clusters_df, "Cluster"),
    errors="coerce"
).astype(int)

# Tengo solo i due gruppi richiesti
clusters_df = clusters_df[
    clusters_df["Oncotree_Group"].isin(oncotree_groups.keys())
].copy()

# Se non esiste Included_Oncotree_Codes, la creo
if "Included_Oncotree_Codes" not in clusters_df.columns:
    clusters_df["Included_Oncotree_Codes"] = clusters_df["Oncotree_Group"].map(group_labels)

# Se Rank_in_cluster non esiste, lo ricostruisco da HubScore
if "Rank_in_cluster" not in clusters_df.columns:

    if "HubScore" not in clusters_df.columns:
        raise ValueError(
            "Nel file dei cluster non trovo né Rank_in_cluster né HubScore. "
            "Devi prima creare i cluster Leiden con le metriche hub."
        )

    clusters_df["HubScore"] = pd.to_numeric(
        get_single_column(clusters_df, "HubScore"),
        errors="coerce"
    )

    clusters_df = clusters_df.sort_values(
        ["Oncotree_Group", "Cluster", "HubScore"],
        ascending=[True, True, False]
    )

    clusters_df["Rank_in_cluster"] = (
            clusters_df
            .groupby(["Oncotree_Group", "Cluster"])
            .cumcount() + 1
    )

clusters_df["Rank_in_cluster"] = pd.to_numeric(
    get_single_column(clusters_df, "Rank_in_cluster"),
    errors="coerce"
)

if "HubScore" in clusters_df.columns:
    clusters_df["HubScore"] = pd.to_numeric(
        get_single_column(clusters_df, "HubScore"),
        errors="coerce"
    )

clusters_df = clusters_df.sort_values(
    ["Oncotree_Group", "Cluster", "Rank_in_cluster"]
).reset_index(drop=True)

print("\nLeiden clusters loaded from:")
print(clusters_path)

print("\nOncotree groups in Leiden cluster file:")
print(sorted(clusters_df["Oncotree_Group"].unique()))

print("\nDuplicate columns in clusters_df:")
print(clusters_df.columns[clusters_df.columns.duplicated()].tolist())


# ============================================================
# 6. Helper functions
# ============================================================

def get_top_hub_genes_text(clusters_sub_df, cluster_id, max_genes=10):
    sub = clusters_sub_df[
        clusters_sub_df["Cluster"] == cluster_id
        ].copy()

    sub = sub.sort_values("Rank_in_cluster")

    labels = []

    for _, row in sub.head(max_genes).iterrows():
        gene = row["Gene"]

        if pd.notna(row.get("Rank_in_cluster", np.nan)):
            rank = int(row["Rank_in_cluster"])
        else:
            rank = len(labels) + 1

        if "HubScore" in sub.columns and pd.notna(row.get("HubScore", np.nan)):
            labels.append(f"{rank}. {gene} ({row['HubScore']:.3f})")
        else:
            labels.append(f"{rank}. {gene}")

    if len(sub) > max_genes:
        labels.append(f"... +{len(sub) - max_genes} genes")

    return " | ".join(labels)


def safe_group_filename(text):
    text = str(text)
    text = text.replace("+", "_")
    text = text.replace("/", "_")
    text = text.replace(" ", "_")
    return text


# ============================================================
# 7. Kaplan-Meier by Oncotree Group and Leiden cluster
# ============================================================

min_patients_per_group = 10
min_events_per_group = 5

survival_results_all = []

available_groups = sorted(
    set(clusters_df["Oncotree_Group"].unique())
    & set(clinical["Oncotree_Group"].unique())
)

print("\nOncotree groups analyzed for KM:")
print(available_groups)

for group_name in available_groups:

    readable_group_name = group_labels.get(group_name, group_name)
    codes = oncotree_groups[group_name]

    print("\n" + "#" * 100)
    print(f"ONCOTREE GROUP: {group_name} ({readable_group_name})")
    print("#" * 100)

    clusters_sub = clusters_df[
        clusters_df["Oncotree_Group"] == group_name
        ].copy()

    clinical_sub = clinical[
        clinical["Oncotree_Group"] == group_name
        ].copy()

    df_network_sub = df[
        (df["ONCOTREE_CODE"].isin(codes))
        & (df["PATIENT_ID"].isin(valid_patients_global))
        ].copy()

    df_network_sub = collapse_duplicate_columns(df_network_sub)

    valid_patients_group = set(
        df_merged_oncotree.loc[
            df_merged_oncotree["ONCOTREE_CODE"].isin(codes),
            "PATIENT_ID"
        ]
        .dropna()
        .astype(str)
        .unique()
    )

    clinical_sub = clinical_sub[
        clinical_sub["PATIENT_ID"].isin(valid_patients_group)
    ].copy()

    df_network_sub = df_network_sub[
        df_network_sub["PATIENT_ID"].isin(valid_patients_group)
    ].copy()

    print(f"Patients in network for {group_name}: {len(valid_patients_group)}")
    print(f"Patients with survival data for {group_name}: {clinical_sub['PATIENT_ID'].nunique()}")
    print(f"Mutation rows for {group_name}: {df_network_sub.shape[0]}")

    cluster_gene_dict = (
        clusters_sub
        .groupby("Cluster")["Gene"]
        .apply(list)
        .to_dict()
    )

    for cluster, genes in cluster_gene_dict.items():
        print(f"Cluster {cluster}: {len(genes)} genes")
        print(
            "Top hubs:",
            get_top_hub_genes_text(
                clusters_sub_df=clusters_sub,
                cluster_id=cluster,
                max_genes=10
            )
        )

    for cluster_id, genes in cluster_gene_dict.items():

        genes = set([str(g) for g in genes])

        top_hub_genes_text = get_top_hub_genes_text(
            clusters_sub_df=clusters_sub,
            cluster_id=cluster_id,
            max_genes=10
        )

        gene_series = get_single_column(df_network_sub, "Gene").astype(str)

        mask_cluster_genes = gene_series.isin(genes)

        patients_with_cluster = (
            df_network_sub.loc[mask_cluster_genes, "PATIENT_ID"]
            .dropna()
            .astype(str)
            .unique()
        )

        surv_df = clinical_sub.copy()

        surv_df["has_cluster"] = (
            surv_df["PATIENT_ID"]
            .isin(patients_with_cluster)
            .astype(int)
        )

        n_cluster = int(surv_df["has_cluster"].sum())
        n_no_cluster = int((surv_df["has_cluster"] == 0).sum())

        events_cluster = int(
            surv_df.loc[
                surv_df["has_cluster"] == 1,
                "event"
            ].sum()
        )

        events_no_cluster = int(
            surv_df.loc[
                surv_df["has_cluster"] == 0,
                "event"
            ].sum()
        )

        print("\n" + "=" * 80)
        print(f"Oncotree Group: {group_name} ({readable_group_name})")
        print(f"Leiden Cluster: {cluster_id}")
        print(f"Genes in cluster: {len(genes)}")
        print(f"Top hub genes: {top_hub_genes_text}")
        print(f"Patients with cluster mutation: {n_cluster}")
        print(f"Patients without cluster mutation: {n_no_cluster}")
        print(f"Events with cluster mutation: {events_cluster}")
        print(f"Events without cluster mutation: {events_no_cluster}")

        if (
                n_cluster < min_patients_per_group
                or n_no_cluster < min_patients_per_group
                or events_cluster < min_events_per_group
                or events_no_cluster < min_events_per_group
        ):
            print("Skipped: too few patients or events.")
            continue

        group1 = surv_df[surv_df["has_cluster"] == 1]
        group0 = surv_df[surv_df["has_cluster"] == 0]

        lr = logrank_test(
            group1["OS_MONTHS"],
            group0["OS_MONTHS"],
            event_observed_A=group1["event"],
            event_observed_B=group0["event"]
        )

        cox_df = surv_df[
            ["OS_MONTHS", "event", "has_cluster"]
        ].copy()

        cph = CoxPHFitter()
        cph.fit(
            cox_df,
            duration_col="OS_MONTHS",
            event_col="event"
        )

        hr = np.exp(cph.params_["has_cluster"])
        p_cox = cph.summary.loc["has_cluster", "p"]

        survival_results_all.append({
            "Oncotree_Group": group_name,
            "Included_Oncotree_Codes": readable_group_name,
            "Cluster": cluster_id,
            "Num_genes": len(genes),
            "Patients_with_cluster": n_cluster,
            "Events_with_cluster": events_cluster,
            "Patients_without_cluster": n_no_cluster,
            "Events_without_cluster": events_no_cluster,
            "Logrank_p_value": lr.p_value,
            "Hazard_Ratio": hr,
            "Cox_p_value": p_cox,
            "Top_hub_genes": top_hub_genes_text,
            "Genes": ", ".join(sorted(genes))
        })

        kmf = KaplanMeierFitter()

        plt.figure(figsize=(11, 6))

        kmf.fit(
            group1["OS_MONTHS"],
            event_observed=group1["event"],
            label=f"{readable_group_name} | Cluster {cluster_id} mutated"
        )
        kmf.plot_survival_function()

        kmf.fit(
            group0["OS_MONTHS"],
            event_observed=group0["event"],
            label=f"{readable_group_name} | Cluster {cluster_id} not mutated"
        )
        kmf.plot_survival_function()

        plt.title(
            f"Kaplan-Meier Survival Curve - {readable_group_name} - Leiden Cluster {cluster_id}\n"
            f"HR={hr:.2f}, Cox p={p_cox:.2e}, Log-rank p={lr.p_value:.2e}\n"
            f"Top hub genes: {top_hub_genes_text}",
            fontsize=9
        )

        plt.xlabel("Overall Survival Months")
        plt.ylabel("Survival probability")
        plt.grid(alpha=0.3)
        plt.tight_layout()

        fig_file = (
                survival_output_dir
                / f"KM_{safe_group_filename(group_name)}_cluster_{cluster_id}.png"
        )

        plt.savefig(fig_file, dpi=300, bbox_inches="tight")
        plt.show()

        print(f"KM plot salvato in: {fig_file}")


# ============================================================
# 8. Final results table
# ============================================================

survival_results_groups = pd.DataFrame(survival_results_all)

if len(survival_results_groups) > 0:
    survival_results_groups = survival_results_groups.sort_values(
        ["Oncotree_Group", "Logrank_p_value"]
    ).reset_index(drop=True)

    output_survival_file = (
            survival_output_dir
            / "survival_results_by_oncotree_group_and_leiden_cluster.csv"
    )

    survival_results_groups.to_csv(
        output_survival_file,
        index=False
    )

    print("\nFinal survival results by Oncotree group and Leiden cluster:")

    display(survival_results_groups)

    print("\nSurvival results salvati in:")
    print(output_survival_file)

else:
    survival_results_groups = pd.DataFrame()
    print("\nNo cluster had enough patients/events for survival analysis.")


# ============================================================
# ALIAS OPZIONALE PER COMPATIBILITÀ
# ============================================================

survival_results_oncotree = survival_results_groups
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from pathlib import Path
from scipy.stats import chi2_contingency, fisher_exact, mannwhitneyu
from lifelines import KaplanMeierFitter, CoxPHFitter
from lifelines.statistics import logrank_test


# ============================================================
# 0. Paths
# ============================================================

data_dir = Path("data_lung")

filtered_path = data_dir / "filtered_data_LUNG.csv"
merged_path = data_dir / "merged_mutations_cna_sv.csv"

clusters_path = (
        data_dir
        / "networks_by_oncotree_groups"
        / "final_leiden_clusters_with_hub_metrics"
        / "leiden_clusters_with_hub_metrics_all_oncotree_groups.csv"
)

clinical_output_dir = (
        data_dir
        / "networks_by_oncotree_groups"
        / "clinical_characterization_by_cluster"
)

clinical_output_dir.mkdir(parents=True, exist_ok=True)


# ============================================================
# 1. Gruppi Oncotree
# ============================================================

oncotree_groups = {
    "LUAD_NSCLC": ["LUAD", "NSCLC"],
    "LUSC_NSCLC": ["LUSC", "NSCLC"],
}

group_labels = {
    "LUAD_NSCLC": "LUAD+NSCLC",
    "LUSC_NSCLC": "LUSC+NSCLC",
}


# ============================================================
# 2. Plot settings
# ============================================================

# Colori coerenti per tutto lo script:
# ogni volta che il colore rappresenta lo stato del cluster,
# Not mutated = blu, Mutated = arancione.
MUTATION_COLORS = {
    "Not mutated": "#4C78A8",
    "Mutated": "#F58518",
}

MUTATION_ORDER = ["Not mutated", "Mutated"]
CI_ALPHA = 0.20


# ============================================================
# 3. Helper functions
# ============================================================

def collapse_duplicate_columns(df):
    df = df.copy()

    if not df.columns.duplicated().any():
        return df

    ordered_unique_cols = pd.unique(df.columns)
    collapsed = pd.DataFrame(index=df.index)

    for col in ordered_unique_cols:
        same_name_cols = df.loc[:, df.columns == col]

        if same_name_cols.shape[1] == 1:
            collapsed[col] = same_name_cols.iloc[:, 0]
        else:
            collapsed[col] = same_name_cols.bfill(axis=1).iloc[:, 0]

    return collapsed.copy()


def get_single_column(df, col):
    obj = df[col]

    if isinstance(obj, pd.DataFrame):
        return obj.iloc[:, 0]

    return obj


def safe_group_filename(text):
    text = str(text)
    text = text.replace("+", "_")
    text = text.replace("/", "_")
    text = text.replace(" ", "_")
    return text


def normalize_smoke(value):
    if pd.isna(value):
        return np.nan

    value = str(value).strip().lower()

    if value in ["", "nan", "none", "unknown", "not available"]:
        return np.nan

    if "never" in value:
        return "Never"

    if "former" in value or "current" in value or "smoker" in value:
        return "Former/Current Smoker"

    return np.nan


def normalize_sex(value):
    if pd.isna(value):
        return np.nan

    value = str(value).strip().lower()

    if value == "male":
        return "Male"

    if value == "female":
        return "Female"

    return np.nan


def clinical_association_categorical(surv_df, clinical_col):
    """
    Associazione tra has_cluster e una variabile clinica categorica.
    Esempi: SMOKE_STATUS, SEX, AGE_GROUP.
    """

    tmp = surv_df[["has_cluster", clinical_col]].dropna().copy()

    if tmp.empty:
        return {
            "test": "not_available",
            "p_value": np.nan,
            "table": "",
        }

    tab = pd.crosstab(tmp["has_cluster"], tmp[clinical_col])
    tab = tab.reindex(index=[0, 1], fill_value=0)

    if tab.shape[0] < 2 or tab.shape[1] < 2:
        return {
            "test": "not_enough_variability",
            "p_value": np.nan,
            "table": tab.to_string(),
        }

    try:
        chi2, p_chi, dof, expected = chi2_contingency(tab)

        if tab.shape == (2, 2) and (expected < 5).any():
            oddsratio, p_fisher = fisher_exact(tab.values)
            return {
                "test": "Fisher exact",
                "p_value": p_fisher,
                "table": tab.to_string(),
            }

        return {
            "test": "Chi-square",
            "p_value": p_chi,
            "table": tab.to_string(),
        }

    except Exception as e:
        return {
            "test": f"failed: {e}",
            "p_value": np.nan,
            "table": tab.to_string(),
        }


def clinical_association_age(surv_df):
    """
    Confronta l'età tra pazienti mutated e not mutated.
    Usa Mann-Whitney U test.
    """

    tmp = surv_df[["has_cluster", "AGE"]].dropna().copy()

    age_cluster = tmp.loc[tmp["has_cluster"] == 1, "AGE"]
    age_no_cluster = tmp.loc[tmp["has_cluster"] == 0, "AGE"]

    if len(age_cluster) < 3 or len(age_no_cluster) < 3:
        return {
            "test": "not_enough_samples",
            "p_value": np.nan,
            "median_age_cluster": np.nan,
            "median_age_no_cluster": np.nan,
        }

    try:
        stat, p_value = mannwhitneyu(
            age_cluster,
            age_no_cluster,
            alternative="two-sided",
        )

        return {
            "test": "Mann-Whitney U",
            "p_value": p_value,
            "median_age_cluster": float(age_cluster.median()),
            "median_age_no_cluster": float(age_no_cluster.median()),
        }

    except Exception as e:
        return {
            "test": f"failed: {e}",
            "p_value": np.nan,
            "median_age_cluster": float(age_cluster.median()),
            "median_age_no_cluster": float(age_no_cluster.median()),
        }


def fit_unadjusted_cox(surv_df):
    """
    Cox univariata:
    Survival ~ has_cluster
    """

    cox_df = surv_df[
        ["OS_MONTHS", "event", "has_cluster"]
    ].dropna().copy()

    if cox_df.shape[0] < 10:
        return {
            "status": "skipped_too_few_samples",
            "n": cox_df.shape[0],
            "hr": np.nan,
            "ci_lower": np.nan,
            "ci_upper": np.nan,
            "p": np.nan,
        }

    if cox_df["has_cluster"].nunique() < 2:
        return {
            "status": "skipped_no_cluster_variability",
            "n": cox_df.shape[0],
            "hr": np.nan,
            "ci_lower": np.nan,
            "ci_upper": np.nan,
            "p": np.nan,
        }

    if cox_df["event"].sum() < 5:
        return {
            "status": "skipped_too_few_events",
            "n": cox_df.shape[0],
            "hr": np.nan,
            "ci_lower": np.nan,
            "ci_upper": np.nan,
            "p": np.nan,
        }

    try:
        cph = CoxPHFitter()
        cph.fit(
            cox_df,
            duration_col="OS_MONTHS",
            event_col="event",
        )

        summary = cph.summary

        return {
            "status": "ok",
            "n": cox_df.shape[0],
            "hr": float(summary.loc["has_cluster", "exp(coef)"]),
            "ci_lower": float(summary.loc["has_cluster", "exp(coef) lower 95%"]),
            "ci_upper": float(summary.loc["has_cluster", "exp(coef) upper 95%"]),
            "p": float(summary.loc["has_cluster", "p"]),
        }

    except Exception as e:
        return {
            "status": f"failed: {e}",
            "n": cox_df.shape[0],
            "hr": np.nan,
            "ci_lower": np.nan,
            "ci_upper": np.nan,
            "p": np.nan,
        }


def fit_adjusted_cox(surv_df):
    """
    Cox multivariata:
    Survival ~ has_cluster + AGE + SMOKE_BINARY + SEX_BINARY
    """

    cox_cols = [
        "OS_MONTHS",
        "event",
        "has_cluster",
        "AGE",
        "SMOKE_BINARY",
        "SEX_BINARY",
    ]

    cox_df = surv_df[cox_cols].dropna().copy()

    if cox_df.shape[0] < 20:
        return {
            "status": "skipped_too_few_complete_cases",
            "n": cox_df.shape[0],
            "hr": np.nan,
            "ci_lower": np.nan,
            "ci_upper": np.nan,
            "p": np.nan,
        }

    if cox_df["has_cluster"].nunique() < 2:
        return {
            "status": "skipped_no_cluster_variability",
            "n": cox_df.shape[0],
            "hr": np.nan,
            "ci_lower": np.nan,
            "ci_upper": np.nan,
            "p": np.nan,
        }

    if cox_df["event"].sum() < 10:
        return {
            "status": "skipped_too_few_events",
            "n": cox_df.shape[0],
            "hr": np.nan,
            "ci_lower": np.nan,
            "ci_upper": np.nan,
            "p": np.nan,
        }

    covariates = ["has_cluster"]

    for cov in ["AGE", "SMOKE_BINARY", "SEX_BINARY"]:
        if cox_df[cov].nunique(dropna=True) > 1:
            covariates.append(cov)

    model_df = cox_df[["OS_MONTHS", "event"] + covariates].copy()

    try:
        cph = CoxPHFitter()
        cph.fit(
            model_df,
            duration_col="OS_MONTHS",
            event_col="event",
        )

        summary = cph.summary

        return {
            "status": "ok",
            "n": model_df.shape[0],
            "hr": float(summary.loc["has_cluster", "exp(coef)"]),
            "ci_lower": float(summary.loc["has_cluster", "exp(coef) lower 95%"]),
            "ci_upper": float(summary.loc["has_cluster", "exp(coef) upper 95%"]),
            "p": float(summary.loc["has_cluster", "p"]),
        }

    except Exception as e:
        return {
            "status": f"failed: {e}",
            "n": model_df.shape[0],
            "hr": np.nan,
            "ci_lower": np.nan,
            "ci_upper": np.nan,
            "p": np.nan,
        }


def get_top_hub_genes_text(clusters_sub_df, cluster_id, max_genes=8):
    sub = clusters_sub_df[
        clusters_sub_df["Cluster"] == cluster_id
        ].copy()

    if "Rank_in_cluster" in sub.columns:
        sub = sub.sort_values("Rank_in_cluster")

    labels = []

    for _, row in sub.head(max_genes).iterrows():
        gene = row["Gene"]

        if "Rank_in_cluster" in sub.columns and pd.notna(row.get("Rank_in_cluster", np.nan)):
            rank = int(row["Rank_in_cluster"])
        else:
            rank = len(labels) + 1

        if "HubScore" in sub.columns and pd.notna(row.get("HubScore", np.nan)):
            labels.append(f"{rank}. {gene} ({row['HubScore']:.3f})")
        else:
            labels.append(f"{rank}. {gene}")

    if len(sub) > max_genes:
        labels.append(f"... +{len(sub) - max_genes} genes")

    return " | ".join(labels)


# ============================================================
# 4. Plot helper functions
# ============================================================

def plot_km(ax, surv_df, lr_p):
    """
    Kaplan-Meier con intervallo di confidenza 95%.
    Colori coerenti:
    - Not mutated: blu
    - Mutated: arancione
    """

    kmf = KaplanMeierFitter()

    plotted = False

    for status_label in MUTATION_ORDER:
        status_value = 1 if status_label == "Mutated" else 0

        sub = surv_df[
            surv_df["has_cluster"] == status_value
            ].copy()

        if sub.empty:
            continue

        kmf.fit(
            durations=sub["OS_MONTHS"],
            event_observed=sub["event"],
            label=status_label,
        )

        kmf.plot_survival_function(
            ax=ax,
            ci_show=True,
            ci_alpha=CI_ALPHA,
            color=MUTATION_COLORS[status_label],
            linewidth=2,
        )

        plotted = True

    if not plotted:
        ax.text(0.5, 0.5, "Not enough data", ha="center", va="center")
        ax.axis("off")
        return

    ax.set_title(
        f"Overall survival\nKaplan-Meier, log-rank p={lr_p:.2e}"
    )
    ax.set_xlabel("Overall survival months")
    ax.set_ylabel("Survival probability")
    ax.set_ylim(0, 1.05)
    ax.grid(alpha=0.25)
    ax.legend(title="Cluster status")


def plot_cluster_cox_forest(ax, unadjusted_cox, adjusted_cox):
    """
    Forest plot locale per singolo cluster.
    Mostra HR e 95% CI della Cox univariata e aggiustata.
    """

    rows = []

    if unadjusted_cox["status"] == "ok":
        rows.append({
            "label": "Unadjusted",
            "hr": unadjusted_cox["hr"],
            "ci_lower": unadjusted_cox["ci_lower"],
            "ci_upper": unadjusted_cox["ci_upper"],
            "p": unadjusted_cox["p"],
        })

    if adjusted_cox["status"] == "ok":
        rows.append({
            "label": "Adjusted\nsmoke+sex+age",
            "hr": adjusted_cox["hr"],
            "ci_lower": adjusted_cox["ci_lower"],
            "ci_upper": adjusted_cox["ci_upper"],
            "p": adjusted_cox["p"],
        })

    if len(rows) == 0:
        status_text = (
            "Cox model not available\n"
            f"Unadjusted: {unadjusted_cox['status']}\n"
            f"Adjusted: {adjusted_cox['status']}"
        )
        ax.text(0.5, 0.5, status_text, ha="center", va="center", fontsize=9)
        ax.set_title("Cox survival analysis")
        ax.axis("off")
        return

    forest_df = pd.DataFrame(rows)
    y_pos = np.arange(forest_df.shape[0])

    hr = forest_df["hr"].values
    ci_lower = forest_df["ci_lower"].values
    ci_upper = forest_df["ci_upper"].values

    xerr = np.vstack([
        hr - ci_lower,
        ci_upper - hr,
    ])

    ax.errorbar(
        hr,
        y_pos,
        xerr=xerr,
        fmt="o",
        capsize=4,
        color="black",
    )

    ax.axvline(
        x=1,
        linestyle="--",
        color="gray",
        linewidth=1,
    )

    ax.set_yticks(y_pos)
    ax.set_yticklabels(forest_df["label"])
    ax.set_xlabel("Hazard Ratio, 95% CI")
    ax.set_title("Cox survival analysis")

    min_x = np.nanmin(ci_lower)
    max_x = np.nanmax(ci_upper)

    if pd.notna(min_x) and pd.notna(max_x):
        ax.set_xlim(
            left=max(0, min_x * 0.75),
            right=max_x * 1.55,
        )

    ax.grid(axis="x", alpha=0.25)

    for i, row in forest_df.iterrows():
        ax.text(
            row["ci_upper"] * 1.03,
            i,
            f"HR={row['hr']:.2f}\n"
            f"95% CI {row['ci_lower']:.2f}-{row['ci_upper']:.2f}\n"
            f"p={row['p']:.2e}",
            va="center",
            fontsize=8,
        )


def plot_clinical_composition_by_mutation_status(
        ax,
        surv_df,
        clinical_col,
        title,
        p_value=None,
):
    """
    Barplot percentuale per variabili cliniche categoriche.

    x = categorie cliniche, es. Never / Former-Current Smoker
    y = percentuale di pazienti dentro quella categoria clinica
    colore = mutation status del cluster

    In questo modo il colore indica sempre la stessa cosa:
    blu = Not mutated, arancione = Mutated.
    """

    tmp = surv_df[
        [clinical_col, "has_cluster_label"]
    ].dropna().copy()

    if tmp.empty:
        ax.text(0.5, 0.5, "Not enough data", ha="center", va="center")
        ax.set_title(title)
        ax.axis("off")
        return

    if tmp[clinical_col].nunique() < 2:
        ax.text(0.5, 0.5, "Not enough variability", ha="center", va="center")
        ax.set_title(title)
        ax.axis("off")
        return

    tab = pd.crosstab(
        tmp[clinical_col],
        tmp["has_cluster_label"],
        normalize="index",
    ) * 100

    for status_label in MUTATION_ORDER:
        if status_label not in tab.columns:
            tab[status_label] = 0

    tab = tab[MUTATION_ORDER]

    counts = pd.crosstab(
        tmp[clinical_col],
        tmp["has_cluster_label"],
    )

    for status_label in MUTATION_ORDER:
        if status_label not in counts.columns:
            counts[status_label] = 0

    counts = counts[MUTATION_ORDER]

    x = np.arange(tab.shape[0])
    width = 0.38

    bars_not = ax.bar(
        x - width / 2,
        tab["Not mutated"],
        width=width,
        label="Not mutated",
        color=MUTATION_COLORS["Not mutated"],
    )

    bars_mut = ax.bar(
        x + width / 2,
        tab["Mutated"],
        width=width,
        label="Mutated",
        color=MUTATION_COLORS["Mutated"],
    )

    ax.set_xticks(x)
    ax.set_xticklabels(tab.index, rotation=20, ha="right")
    ax.set_ylabel("Patients (%)")
    ax.set_ylim(0, 100)

    if p_value is not None and pd.notna(p_value):
        ax.set_title(f"{title}\np={p_value:.2e}")
    else:
        ax.set_title(title)

    ax.legend(title="Cluster status", fontsize=8)
    ax.grid(axis="y", alpha=0.25)

    # Etichette: percentuale + n assoluto.
    for bar_container, status_label in [
        (bars_not, "Not mutated"),
        (bars_mut, "Mutated"),
    ]:
        for bar, category in zip(bar_container, tab.index):
            height = bar.get_height()
            n_abs = int(counts.loc[category, status_label])

            if height > 0:
                ax.text(
                    bar.get_x() + bar.get_width() / 2,
                    height + 1.5,
                    f"{height:.1f}%\nn={n_abs}",
                    ha="center",
                    va="bottom",
                    fontsize=8,
                )


def plot_age_distribution(ax, surv_df, p_value=None):
    """
    Boxplot età per cluster status.
    Colori coerenti con Mutated / Not mutated.
    """

    tmp = surv_df[
        ["has_cluster_label", "AGE"]
    ].dropna().copy()

    if tmp.empty or tmp["has_cluster_label"].nunique() < 2:
        ax.text(0.5, 0.5, "Not enough data", ha="center", va="center")
        ax.set_title("Age distribution")
        ax.axis("off")
        return

    data = []
    labels = []

    for status_label in MUTATION_ORDER:
        values = tmp.loc[
            tmp["has_cluster_label"] == status_label,
            "AGE",
        ].values

        if len(values) > 0:
            data.append(values)
            labels.append(status_label)

    box = ax.boxplot(
        data,
        labels=labels,
        patch_artist=True,
        showfliers=True,
    )

    for patch, label in zip(box["boxes"], labels):
        patch.set_facecolor(MUTATION_COLORS[label])
        patch.set_alpha(0.65)

    for median in box["medians"]:
        median.set_color("black")
        median.set_linewidth(2)

    if p_value is not None and pd.notna(p_value):
        ax.set_title(f"Age distribution\nMann-Whitney p={p_value:.2e}")
    else:
        ax.set_title("Age distribution")

    ax.set_ylabel("Age")
    ax.grid(axis="y", alpha=0.25)

    # Aggiunge n e mediana sotto le categorie.
    xtick_labels = []
    for label, values in zip(labels, data):
        xtick_labels.append(
            f"{label}\nn={len(values)}, median={np.median(values):.1f}"
        )

    ax.set_xticklabels(xtick_labels, rotation=0)


# ============================================================
# 5. Load clinical/mutation data
# ============================================================

df = pd.read_csv(filtered_path, low_memory=False)
df = collapse_duplicate_columns(df)

if "Gene" in df.columns and "Hugo_Symbol" in df.columns:
    df = df.drop(columns=["Gene"])

df = df.rename(columns={
    "Overall Survival (Months)": "OS_MONTHS",
    "Overall Survival Status": "OS_STATUS",
    "Patient_ID": "PATIENT_ID",
    "Hugo_Symbol": "Gene",
    "Oncotree Code": "ONCOTREE_CODE",
})

df = collapse_duplicate_columns(df)

required_cols = {
    "PATIENT_ID",
    "Gene",
    "ONCOTREE_CODE",
    "OS_MONTHS",
    "OS_STATUS",
    "Sex",
    "Current Age",
    "Smoking History (NLP)",
}

missing_cols = required_cols - set(df.columns)

if missing_cols:
    raise ValueError(f"Colonne mancanti nel file clinico: {missing_cols}")

df["PATIENT_ID"] = get_single_column(df, "PATIENT_ID").astype(str)
df["Gene"] = get_single_column(df, "Gene").astype(str)
df["ONCOTREE_CODE"] = get_single_column(df, "ONCOTREE_CODE").astype(str)

df["OS_MONTHS"] = pd.to_numeric(
    get_single_column(df, "OS_MONTHS"),
    errors="coerce",
)

df["event"] = (
    get_single_column(df, "OS_STATUS")
    .astype(str)
    .str.contains("DECEASED", case=False, na=False)
    .astype(int)
)

df["SEX"] = get_single_column(df, "Sex").apply(normalize_sex)

df["SEX_BINARY"] = df["SEX"].map({
    "Female": 0,
    "Male": 1,
})

df["SMOKE_STATUS"] = get_single_column(
    df,
    "Smoking History (NLP)",
).apply(normalize_smoke)

df["SMOKE_BINARY"] = df["SMOKE_STATUS"].map({
    "Never": 0,
    "Former/Current Smoker": 1,
})

df["AGE"] = pd.to_numeric(
    get_single_column(df, "Current Age"),
    errors="coerce",
)

df["AGE_GROUP"] = np.where(
    df["AGE"] >= 65,
    "Age >= 65",
    "Age < 65",
)

df.loc[df["AGE"].isna(), "AGE_GROUP"] = np.nan

df = df.dropna(
    subset=[
        "PATIENT_ID",
        "Gene",
        "ONCOTREE_CODE",
        "OS_MONTHS",
        "event",
    ]
).copy()

print("Clinical/mutation data loaded")
print("Rows:", df.shape[0])
print("Patients:", df["PATIENT_ID"].nunique())

display(
    df[[
        "PATIENT_ID",
        "ONCOTREE_CODE",
        "OS_MONTHS",
        "event",
        "SMOKE_STATUS",
        "SEX",
        "AGE",
    ]]
    .drop_duplicates("PATIENT_ID")
    .head()
)


# ============================================================
# 6. Load merged data to define valid network patients
# ============================================================

df_merged = pd.read_csv(merged_path, low_memory=False)
df_merged = collapse_duplicate_columns(df_merged)

if "Gene" in df_merged.columns and "Hugo_Symbol" in df_merged.columns:
    df_merged = df_merged.drop(columns=["Gene"])

df_merged = df_merged.rename(columns={
    "Patient_ID": "PATIENT_ID",
    "Hugo_Symbol": "Gene",
})

df_merged = collapse_duplicate_columns(df_merged)

if "PATIENT_ID" not in df_merged.columns:
    if "Tumor_Sample_Barcode" in df_merged.columns:
        df_merged["PATIENT_ID"] = (
            df_merged["Tumor_Sample_Barcode"]
            .astype(str)
            .str[:12]
        )
    elif "SAMPLE_ID" in df_merged.columns:
        df_merged["PATIENT_ID"] = (
            df_merged["SAMPLE_ID"]
            .astype(str)
            .str[:12]
        )
    else:
        raise ValueError("Non trovo una colonna paziente in merged.")

if "Gene" not in df_merged.columns:
    raise ValueError("Non trovo Gene/Hugo_Symbol in merged.")

df_merged["PATIENT_ID"] = get_single_column(df_merged, "PATIENT_ID").astype(str)
df_merged["Gene"] = get_single_column(df_merged, "Gene").astype(str)

valid_patients_global = set(
    df_merged["PATIENT_ID"]
    .dropna()
    .astype(str)
    .unique()
)

print("Patients valid from merged/network:", len(valid_patients_global))


# ============================================================
# 7. Clinical table patient-level
# ============================================================

clinical_base = (
    df[
        [
            "PATIENT_ID",
            "ONCOTREE_CODE",
            "OS_MONTHS",
            "event",
            "SMOKE_STATUS",
            "SMOKE_BINARY",
            "SEX",
            "SEX_BINARY",
            "AGE",
            "AGE_GROUP",
        ]
    ]
    .drop_duplicates(["PATIENT_ID", "ONCOTREE_CODE"])
    .copy()
)

clinical_base = clinical_base[
    clinical_base["PATIENT_ID"].isin(valid_patients_global)
].copy()

clinical_group_tables = []

for group_name, codes in oncotree_groups.items():
    readable_group_name = group_labels.get(group_name, group_name)

    clinical_group = clinical_base[
        clinical_base["ONCOTREE_CODE"].isin(codes)
    ].copy()

    clinical_group["Oncotree_Group"] = group_name
    clinical_group["Included_Oncotree_Codes"] = readable_group_name

    clinical_group_tables.append(clinical_group)

clinical = pd.concat(
    clinical_group_tables,
    ignore_index=True,
)

print("\nClinical data by Oncotree group:")

display(
    clinical
    .groupby("Oncotree_Group")
    .agg(
        included_codes=("Included_Oncotree_Codes", "first"),
        n_patients=("PATIENT_ID", "nunique"),
        n_events=("event", "sum"),
        n_with_smoke=("SMOKE_STATUS", lambda x: x.notna().sum()),
        n_with_sex=("SEX", lambda x: x.notna().sum()),
        n_with_age=("AGE", lambda x: x.notna().sum()),
        n_smokers=("SMOKE_BINARY", lambda x: int((x == 1).sum())),
        n_never_smokers=("SMOKE_BINARY", lambda x: int((x == 0).sum())),
    )
)


# ============================================================
# 8. Load Leiden clusters
# ============================================================

clusters_df = pd.read_csv(clusters_path)
clusters_df = collapse_duplicate_columns(clusters_df)

clusters_df = clusters_df.rename(columns={
    "Hugo_Symbol": "Gene",
    "gene": "Gene",
    "cluster": "Cluster",
    "Leiden_cluster": "Cluster",
    "hubscore": "HubScore",
    "rank_in_cluster": "Rank_in_cluster",
    "Oncotree_Code": "Oncotree_Group",
    "ONCOTREE_CODE": "Oncotree_Group",
    "Oncotree Code": "Oncotree_Group",
})

clusters_df = collapse_duplicate_columns(clusters_df)

required_cluster_cols = {
    "Oncotree_Group",
    "Gene",
    "Cluster",
}

missing_cluster_cols = required_cluster_cols - set(clusters_df.columns)

if missing_cluster_cols:
    raise ValueError(
        f"Colonne mancanti nel file cluster Leiden: {missing_cluster_cols}"
    )

clusters_df["Oncotree_Group"] = get_single_column(
    clusters_df,
    "Oncotree_Group",
).astype(str)

clusters_df["Gene"] = get_single_column(
    clusters_df,
    "Gene",
).astype(str)

clusters_df["Cluster"] = pd.to_numeric(
    get_single_column(clusters_df, "Cluster"),
    errors="coerce",
).astype(int)

clusters_df = clusters_df[
    clusters_df["Oncotree_Group"].isin(oncotree_groups.keys())
].copy()

if "Rank_in_cluster" not in clusters_df.columns:
    if "HubScore" in clusters_df.columns:
        clusters_df["HubScore"] = pd.to_numeric(
            get_single_column(clusters_df, "HubScore"),
            errors="coerce",
        )

        clusters_df = clusters_df.sort_values(
            ["Oncotree_Group", "Cluster", "HubScore"],
            ascending=[True, True, False],
        )

        clusters_df["Rank_in_cluster"] = (
                clusters_df
                .groupby(["Oncotree_Group", "Cluster"])
                .cumcount() + 1
        )
    else:
        clusters_df["Rank_in_cluster"] = (
                clusters_df
                .groupby(["Oncotree_Group", "Cluster"])
                .cumcount() + 1
        )

clusters_df["Rank_in_cluster"] = pd.to_numeric(
    get_single_column(clusters_df, "Rank_in_cluster"),
    errors="coerce",
)

if "HubScore" in clusters_df.columns:
    clusters_df["HubScore"] = pd.to_numeric(
        get_single_column(clusters_df, "HubScore"),
        errors="coerce",
    )

print("\nLeiden clusters loaded")

display(
    clusters_df
    .groupby(["Oncotree_Group", "Cluster"])
    .agg(
        n_genes=("Gene", "nunique"),
    )
    .reset_index()
)


# ============================================================
# 9. Clinical characterization + plots
# ============================================================

min_patients_per_group = 10
min_events_per_group = 5

clinical_results_all = []

available_groups = sorted(
    set(clusters_df["Oncotree_Group"].unique())
    & set(clinical["Oncotree_Group"].unique())
)

for group_name in available_groups:

    readable_group_name = group_labels.get(group_name, group_name)
    codes = oncotree_groups[group_name]

    print("\n" + "#" * 100)
    print(f"CLINICAL CHARACTERIZATION: {group_name} ({readable_group_name})")
    print("#" * 100)

    group_output_dir = clinical_output_dir / safe_group_filename(group_name)
    group_output_dir.mkdir(parents=True, exist_ok=True)

    clusters_sub = clusters_df[
        clusters_df["Oncotree_Group"] == group_name
        ].copy()

    clinical_sub = clinical[
        clinical["Oncotree_Group"] == group_name
        ].copy()

    df_network_sub = df[
        (df["ONCOTREE_CODE"].isin(codes))
        & (df["PATIENT_ID"].isin(valid_patients_global))
        ].copy()

    valid_patients_group = set(
        df_network_sub["PATIENT_ID"]
        .dropna()
        .astype(str)
        .unique()
    )

    clinical_sub = clinical_sub[
        clinical_sub["PATIENT_ID"].isin(valid_patients_group)
    ].copy()

    cluster_gene_dict = (
        clusters_sub
        .groupby("Cluster")["Gene"]
        .apply(list)
        .to_dict()
    )

    for cluster_id, genes in cluster_gene_dict.items():

        genes = set([str(g) for g in genes])

        top_hub_genes_text = get_top_hub_genes_text(
            clusters_sub_df=clusters_sub,
            cluster_id=cluster_id,
            max_genes=8,
        )

        patients_with_cluster = (
            df_network_sub.loc[
                df_network_sub["Gene"].astype(str).isin(genes),
                "PATIENT_ID",
            ]
            .dropna()
            .astype(str)
            .unique()
        )

        surv_df = clinical_sub.copy()

        surv_df["has_cluster"] = (
            surv_df["PATIENT_ID"]
            .isin(patients_with_cluster)
            .astype(int)
        )

        surv_df["has_cluster_label"] = surv_df["has_cluster"].map({
            0: "Not mutated",
            1: "Mutated",
        })

        n_cluster = int(surv_df["has_cluster"].sum())
        n_no_cluster = int((surv_df["has_cluster"] == 0).sum())

        events_cluster = int(
            surv_df.loc[surv_df["has_cluster"] == 1, "event"].sum()
        )

        events_no_cluster = int(
            surv_df.loc[surv_df["has_cluster"] == 0, "event"].sum()
        )

        print("\n" + "=" * 80)
        print(f"{readable_group_name} | Cluster {cluster_id}")
        print(f"Genes: {len(genes)}")
        print(f"Top hubs: {top_hub_genes_text}")
        print(f"Mutated patients: {n_cluster} | events: {events_cluster}")
        print(f"Not mutated patients: {n_no_cluster} | events: {events_no_cluster}")

        if (
                n_cluster < min_patients_per_group
                or n_no_cluster < min_patients_per_group
                or events_cluster < min_events_per_group
                or events_no_cluster < min_events_per_group
        ):
            print("Skipped: too few patients or events.")

            clinical_results_all.append({
                "Oncotree_Group": group_name,
                "Included_Oncotree_Codes": readable_group_name,
                "Cluster": cluster_id,
                "Num_genes": len(genes),
                "Patients_mutated": n_cluster,
                "Events_mutated": events_cluster,
                "Patients_not_mutated": n_no_cluster,
                "Events_not_mutated": events_no_cluster,
                "Status": "skipped_too_few_patients_or_events",
                "Top_hub_genes": top_hub_genes_text,
                "Genes": ", ".join(sorted(genes)),
            })

            continue

        group1 = surv_df[surv_df["has_cluster"] == 1]
        group0 = surv_df[surv_df["has_cluster"] == 0]

        lr = logrank_test(
            group1["OS_MONTHS"],
            group0["OS_MONTHS"],
            event_observed_A=group1["event"],
            event_observed_B=group0["event"],
        )

        smoke_assoc = clinical_association_categorical(
            surv_df,
            "SMOKE_STATUS",
        )

        sex_assoc = clinical_association_categorical(
            surv_df,
            "SEX",
        )

        age_group_assoc = clinical_association_categorical(
            surv_df,
            "AGE_GROUP",
        )

        age_cont_assoc = clinical_association_age(surv_df)

        unadjusted_cox = fit_unadjusted_cox(surv_df)
        adjusted_cox = fit_adjusted_cox(surv_df)

        print("Smoke association:", smoke_assoc["test"], "| p =", smoke_assoc["p_value"])
        print("Sex association:", sex_assoc["test"], "| p =", sex_assoc["p_value"])
        print("Age group association:", age_group_assoc["test"], "| p =", age_group_assoc["p_value"])
        print("Age continuous association:", age_cont_assoc["test"], "| p =", age_cont_assoc["p_value"])
        print(
            "Unadjusted Cox:",
            unadjusted_cox["status"],
            "| HR =",
            unadjusted_cox["hr"],
            "| 95% CI =",
            unadjusted_cox["ci_lower"],
            "-",
            unadjusted_cox["ci_upper"],
            "| p =",
            unadjusted_cox["p"],
        )
        print(
            "Adjusted Cox:",
            adjusted_cox["status"],
            "| HR =",
            adjusted_cox["hr"],
            "| 95% CI =",
            adjusted_cox["ci_lower"],
            "-",
            adjusted_cox["ci_upper"],
            "| p =",
            adjusted_cox["p"],
        )

        clinical_results_all.append({
            "Oncotree_Group": group_name,
            "Included_Oncotree_Codes": readable_group_name,
            "Cluster": cluster_id,
            "Num_genes": len(genes),

            "Patients_mutated": n_cluster,
            "Events_mutated": events_cluster,
            "Patients_not_mutated": n_no_cluster,
            "Events_not_mutated": events_no_cluster,

            "Logrank_p_value": lr.p_value,

            "Unadjusted_Cox_status": unadjusted_cox["status"],
            "Unadjusted_Cox_n": unadjusted_cox["n"],
            "Unadjusted_HR": unadjusted_cox["hr"],
            "Unadjusted_HR_CI_lower": unadjusted_cox["ci_lower"],
            "Unadjusted_HR_CI_upper": unadjusted_cox["ci_upper"],
            "Unadjusted_Cox_p_value": unadjusted_cox["p"],

            "Adjusted_Cox_status": adjusted_cox["status"],
            "Adjusted_Cox_n": adjusted_cox["n"],
            "Adjusted_HR": adjusted_cox["hr"],
            "Adjusted_HR_CI_lower": adjusted_cox["ci_lower"],
            "Adjusted_HR_CI_upper": adjusted_cox["ci_upper"],
            "Adjusted_Cox_p_value": adjusted_cox["p"],

            "Smoke_association_test": smoke_assoc["test"],
            "Smoke_association_p_value": smoke_assoc["p_value"],
            "Smoke_table": smoke_assoc["table"],

            "Sex_association_test": sex_assoc["test"],
            "Sex_association_p_value": sex_assoc["p_value"],
            "Sex_table": sex_assoc["table"],

            "Age_group_association_test": age_group_assoc["test"],
            "Age_group_association_p_value": age_group_assoc["p_value"],
            "Age_group_table": age_group_assoc["table"],

            "Age_continuous_test": age_cont_assoc["test"],
            "Age_continuous_p_value": age_cont_assoc["p_value"],
            "Median_age_mutated": age_cont_assoc["median_age_cluster"],
            "Median_age_not_mutated": age_cont_assoc["median_age_no_cluster"],

            "Status": "ok",
            "Top_hub_genes": top_hub_genes_text,
            "Genes": ", ".join(sorted(genes)),
        })

        # ====================================================
        # 6-panel plot per cluster
        # ====================================================

        fig, axes = plt.subplots(
            2,
            3,
            figsize=(20, 11),
        )

        if adjusted_cox["status"] == "ok":
            survival_text = (
                f"Adjusted HR={adjusted_cox['hr']:.2f} "
                f"[95% CI {adjusted_cox['ci_lower']:.2f}-{adjusted_cox['ci_upper']:.2f}], "
                f"p={adjusted_cox['p']:.2e}"
            )
        else:
            survival_text = f"Adjusted Cox: {adjusted_cox['status']}"

        fig.suptitle(
            f"{readable_group_name} | Leiden Cluster {cluster_id}\n"
            f"Top hubs: {top_hub_genes_text}\n"
            f"{survival_text}",
            fontsize=12,
        )

        plot_km(
            ax=axes[0, 0],
            surv_df=surv_df,
            lr_p=lr.p_value,
        )

        plot_cluster_cox_forest(
            ax=axes[0, 1],
            unadjusted_cox=unadjusted_cox,
            adjusted_cox=adjusted_cox,
        )

        plot_clinical_composition_by_mutation_status(
            ax=axes[0, 2],
            surv_df=surv_df,
            clinical_col="SMOKE_STATUS",
            title="Smoking history",
            p_value=smoke_assoc["p_value"],
        )

        plot_clinical_composition_by_mutation_status(
            ax=axes[1, 0],
            surv_df=surv_df,
            clinical_col="SEX",
            title="Sex",
            p_value=sex_assoc["p_value"],
        )

        plot_clinical_composition_by_mutation_status(
            ax=axes[1, 1],
            surv_df=surv_df,
            clinical_col="AGE_GROUP",
            title="Age group",
            p_value=age_group_assoc["p_value"],
        )

        plot_age_distribution(
            ax=axes[1, 2],
            surv_df=surv_df,
            p_value=age_cont_assoc["p_value"],
        )

        plt.tight_layout(rect=[0, 0, 1, 0.88])

        fig_file = (
                group_output_dir
                / f"clinical_characterization_{safe_group_filename(group_name)}_cluster_{cluster_id}.png"
        )

        plt.savefig(fig_file, dpi=300, bbox_inches="tight")
        plt.show()

        print("Saved plot:", fig_file)


# ============================================================
# 10. Save final clinical characterization table
# ============================================================

clinical_results = pd.DataFrame(clinical_results_all)

output_file = clinical_output_dir / "clinical_characterization_results_by_cluster.csv"

clinical_results.to_csv(
    output_file,
    index=False,
)

print("\nClinical characterization results saved in:")
print(output_file)

display(clinical_results)


# ============================================================
# 11. Forest plot finale degli HR aggiustati
# ============================================================

forest_df = clinical_results.copy()

required_forest_cols = {
    "Status",
    "Adjusted_Cox_status",
    "Adjusted_HR",
    "Adjusted_HR_CI_lower",
    "Adjusted_HR_CI_upper",
}

if required_forest_cols.issubset(forest_df.columns):
    forest_df = forest_df[
        (forest_df["Status"] == "ok")
        & (forest_df["Adjusted_Cox_status"] == "ok")
        & forest_df["Adjusted_HR"].notna()
        & forest_df["Adjusted_HR_CI_lower"].notna()
        & forest_df["Adjusted_HR_CI_upper"].notna()
        ].copy()
else:
    forest_df = pd.DataFrame()

if forest_df.empty:
    print("No valid adjusted Cox results available for final forest plot.")

else:
    for group_name in forest_df["Oncotree_Group"].unique():

        sub = forest_df[
            forest_df["Oncotree_Group"] == group_name
            ].copy()

        readable_group_name = group_labels.get(group_name, group_name)
        sub = sub.sort_values("Adjusted_HR")

        labels = [
            f"Cluster {int(row.Cluster)}"
            for _, row in sub.iterrows()
        ]

        y_pos = np.arange(len(sub))

        hr = sub["Adjusted_HR"].values
        ci_lower = sub["Adjusted_HR_CI_lower"].values
        ci_upper = sub["Adjusted_HR_CI_upper"].values

        xerr = np.vstack([
            hr - ci_lower,
            ci_upper - hr,
        ])

        plt.figure(figsize=(10, max(5, len(sub) * 0.60)))

        plt.errorbar(
            hr,
            y_pos,
            xerr=xerr,
            fmt="o",
            capsize=4,
            color="black",
        )

        plt.axvline(
            x=1,
            linestyle="--",
            color="gray",
            linewidth=1,
        )

        plt.yticks(
            y_pos,
            labels,
        )

        plt.xlabel("Adjusted Hazard Ratio, 95% CI")
        plt.title(
            f"Adjusted Cox Forest Plot - {readable_group_name}\n"
            "Model: Survival ~ Cluster altered + Smoke + Sex + Age"
        )

        plt.grid(axis="x", alpha=0.3)
        plt.tight_layout()

        forest_file = (
                clinical_output_dir
                / f"forest_plot_adjusted_HR_{safe_group_filename(group_name)}.png"
        )

        plt.savefig(forest_file, dpi=300, bbox_inches="tight")
        plt.show()

        print("Saved forest plot:", forest_file)


# ============================================================
# 12. Summary plot: -log10 p-values
# ============================================================

summary_df = clinical_results.copy()

if "Status" in summary_df.columns:
    summary_df = summary_df[
        summary_df["Status"] == "ok"
        ].copy()
else:
    summary_df = pd.DataFrame()

if summary_df.empty:
    print("No valid results available for summary p-value plot.")

else:
    for group_name in summary_df["Oncotree_Group"].unique():

        sub = summary_df[
            summary_df["Oncotree_Group"] == group_name
            ].copy()

        readable_group_name = group_labels.get(group_name, group_name)

        plot_rows = []

        for _, row in sub.iterrows():
            cluster_label = f"Cluster {int(row['Cluster'])}"

            p_values = {
                "Smoke": row.get("Smoke_association_p_value", np.nan),
                "Sex": row.get("Sex_association_p_value", np.nan),
                "Age group": row.get("Age_group_association_p_value", np.nan),
                "Age continuous": row.get("Age_continuous_p_value", np.nan),
                "Survival adjusted Cox": row.get("Adjusted_Cox_p_value", np.nan),
            }

            for variable, p_value in p_values.items():
                if pd.notna(p_value) and p_value > 0:
                    plot_rows.append({
                        "Cluster": cluster_label,
                        "Variable": variable,
                        "minus_log10_p": -np.log10(p_value),
                    })

        plot_df = pd.DataFrame(plot_rows)

        if plot_df.empty:
            print(f"No valid p-values for {readable_group_name}.")
            continue

        pivot = plot_df.pivot(
            index="Cluster",
            columns="Variable",
            values="minus_log10_p",
        )

        pivot = pivot.sort_index()

        plt.figure(figsize=(11, max(5, pivot.shape[0] * 0.60)))

        plt.imshow(
            pivot.values,
            aspect="auto",
        )

        plt.colorbar(label="-log10(p-value)")

        plt.xticks(
            ticks=np.arange(pivot.shape[1]),
            labels=pivot.columns,
            rotation=30,
            ha="right",
        )

        plt.yticks(
            ticks=np.arange(pivot.shape[0]),
            labels=pivot.index,
        )

        plt.title(
            f"Clinical association summary - {readable_group_name}"
        )

        plt.tight_layout()

        summary_file = (
                clinical_output_dir
                / f"summary_clinical_association_pvalues_{safe_group_filename(group_name)}.png"
        )

        plt.savefig(summary_file, dpi=300, bbox_inches="tight")
        plt.show()

        print("Saved summary plot:", summary_file)
```
